# Exploratory Analysis of NASA C‑MAPSS FD001
This notebook loads the training data, shows basic statistics, identifies constant/low‑variance sensors, and visualises a few engine trajectories.

```python
import os, pandas as pd, matplotlib.pyplot as plt, seaborn as sns
%matplotlib inline

# Path to data loader module
import sys
sys.path.append(r'c:/Users/samjo/Downloads/simulation/Jet-Engine-Failure-Predictor/src')
from data_loader import load_dataset

# Load FD001 training data
df = load_dataset('FD001', kind='train')
print('Shape:', df.shape)
print(df.head())
```

```python
# Basic statistics
print(df.describe().T[['mean','std','min','max']])
```

```python
# Identify constant sensors (zero variance)
sensor_cols = [c for c in df.columns if c.startswith('sensor_')]
const_sensors = [c for c in sensor_cols if df[c].std() == 0]
print('Constant sensors:', const_sensors)
# Low‑variance sensors (std < 0.01)
lowvar_sensors = [c for c in sensor_cols if df[c].std() < 0.01]
print('Low‑variance sensors (<0.01):', lowvar_sensors)
```

```python
# Plot sensor trajectories for a few engines
engine_ids = df['engine_id'].unique()[:3]  # first three engines
for eng in engine_ids:
    eng_df = df[df['engine_id']==eng]
    plt.figure(figsize=(10,4))
    for sensor in sensor_cols[:5]:  # first 5 sensors
        plt.plot(eng_df['cycle'], eng_df[sensor], label=sensor)
    plt.title(f'Engine {eng} – first 5 sensor trajectories')
    plt.xlabel('Cycle')
    plt.ylabel('Sensor reading')
    plt.legend(loc='upper right')
    plt.tight_layout()
    plt.show()
```
